# Mood Check-In EDA

```python
import glob
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
# import seaborn as sns

# load most recent export
files = sorted(Path("../exports").glob("mood-checkins-*.csv"))
CSV = files[-1]
print(f"Loading: {CSV}")
```

```python
df = pd.read_csv(CSV, parse_dates=["timestamp"])
df.head()
```

```python
df.head()
```

## Missing values

```python
null_counts = df.isnull().sum()
null_pct = (df.isnull().mean() * 100).round(1)
pd.DataFrame({"null_count": null_counts, "null_%": null_pct}).query("null_count > 0").sort_values("null_%", ascending=False)
```

## Categorical columns

```python
# mood is semicolon-separated — explode into one tag per row
mood_counts = df["mood"].dropna().str.split(";").explode().str.strip().value_counts()
mood_counts
```

```python
cat_cols = ["energy", "sleepQuality", "workLocation", "workPace", "socialized", "periodPhase", "activityType"]
for col in cat_cols:
    if col in df.columns:
        print(f"\n--- {col} ---")
        print(df[col].value_counts(dropna=False))
```

```python
num_cols = ["sleepHours", "activeMinutes", "extraWorkMinutes"]
df[num_cols].describe()
```
